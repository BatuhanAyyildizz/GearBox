```python
from sklearn.model_selection import train_test_split
from lightgbm import LGBMRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Define features and target
target_column = "BRS.T35_T_GBX_Avg"
features = df.drop(columns=["time", target_column])  # Drop time and target column
target = df[target_column]

# Split data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

# Train LightGBM model
lgb_model = LGBMRegressor()
lgb_model.fit(X_train, y_train)

# Make predictions
y_pred = lgb_model.predict(X_test)

# Evaluate model
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

# Display evaluation metrics
mae, rmse, r2
```

```python

# Gerekli kütüphaneleri içe aktarın
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import lightgbm as lgb
import matplotlib.pyplot as plt
```

```python

# 1. CSV dosyasını yükleyin (dosya JupyterLab dizininde olmalıdır)
df = pd.read_csv("Interpolated_and_Predicted_Data.csv")

# 2. İlgili sıcaklık sütunlarının ortalamasını alarak yeni bir sütun oluşturun
temperature_columns = [
    "BRS.T35_T_GBX_OIL_1",
    "BRS.T35_T_GBX_OIL_2",
    "BRS.T35_T_GBX_T1_HSS",
    "BRS.T35_T_GBX_T1_IMS",
    "BRS.T35_T_GBX_T3_HSS",
    "BRS.T35_T_GBX_T3_IMS"
]
```

```python
df["BRS.T35_T_GBX_Avg"] = df[temperature_columns].mean(axis=1)

# 3. Orijinal sıcaklık sütunlarını kaldırın
df = df.drop(columns=temperature_columns)
# 4. 'time' sütununu datetime formatına çevirin ve veriyi zaman sırasına göre sıralayın
df["time"] = pd.to_datetime(df["time"])
df = df.sort_values("time")
df
# 5. Veriyi zaman tabanlı olarak ayırın: İlk %75 eğitim, son %25 test
split_index = int(0.75 * len(df))
train_df = df.iloc[:split_index]
test_df = df.iloc[split_index:]

# 6. Modelleme için özellikler (X) ve hedef (y) belirleyin (time ve hedef sütunları hariç)
X_train = train_df.drop(columns=["time", "BRS.T35_T_GBX_Avg"])
y_train = train_df["BRS.T35_T_GBX_Avg"]

X_test = test_df.drop(columns=["time", "BRS.T35_T_GBX_Avg"])
y_test = test_df["BRS.T35_T_GBX_Avg"]

# 7. LightGBM modelini oluşturun ve eğitin
model = lgb.LGBMRegressor(random_state=42)
model.fit(X_train, y_train)

# 8. Test kümesi üzerinde tahmin yapın ve modeli değerlendirin
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred, squared=False)
r2 = r2_score(y_test, y_pred)

print("LightGBM Model Performansı (Test Kümesi):")
print("Mean Absolute Error (MAE):", mae)
print("Root Mean Squared Error (RMSE):", rmse)
print("R2 Score:", r2)
```

```python
# 9. Gerçek vs Tahmin Edilen Değerler Grafiği
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel("Gerçek Değerler")
plt.ylabel("Tahmin Edilen Değerler")
plt.title("Gerçek vs Tahmin Edilen Değerler (LightGBM)")
plt.show()

# 10. Hata (Residual) Grafiği
residuals = y_test - y_pred
plt.figure(figsize=(8,6))
plt.scatter(y_pred, residuals, alpha=0.6)
plt.axhline(0, color='r', linestyle='--')
plt.xlabel("Tahmin Edilen Değerler")
plt.ylabel("Hata (Residual)")
plt.title("Hata Dağılımı (Residual Plot)")
plt.show()
```

```python
import pandas as pd
import lightgbm as lgb
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Interpolated_and_Predicted_Data.csv")

# 2. İlgili sıcaklık sütunlarının ortalamasını alarak yeni bir sütun oluşturun
temperature_columns = [
    "BRS.T35_T_GBX_OIL_1",
    "BRS.T35_T_GBX_OIL_2",
    "BRS.T35_T_GBX_T1_HSS",
    "BRS.T35_T_GBX_T1_IMS",
    "BRS.T35_T_GBX_T3_HSS",
    "BRS.T35_T_GBX_T3_IMS"
]
df["BRS.T35_T_GBX_Avg"] = df[temperature_columns].mean(axis=1)

# 3. Orijinal sıcaklık sütunlarını kaldırın
df = df.drop(columns=temperature_columns)

# 4. 'time' sütununu datetime formatına çevirin ve veriyi zaman sırasına göre sıralayın
df["time"] = pd.to_datetime(df["time"])
df = df.sort_values("time")
```

```python
# 5. Veriyi zamana göre ayırın: İlk %75 eğitim, son %25 test
split_index = int(0.5 * len(df))
train_df = df.iloc[:split_index]
test_df = df.iloc[split_index:]

# 6. Modelleme için özellikler (X) ve hedef (y) belirleyin. 'time' ve hedef sütunu modelden çıkarılır.
X_train = train_df.drop(columns=["time", "BRS.T35_T_GBX_Avg"])
y_train = train_df["BRS.T35_T_GBX_Avg"]

X_test = test_df.drop(columns=["time", "BRS.T35_T_GBX_Avg"])
y_test = test_df["BRS.T35_T_GBX_Avg"]

# 7. LightGBM modelini oluşturun ve eğitin
model = lgb.LGBMRegressor(random_state=42)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

# 9. Model performans metriklerini hesaplayın
mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred, squared=False)
r2 = r2_score(y_test, y_pred)

print("LightGBM Model Performansı (Test Kümesi):")
print("Mean Absolute Error (MAE):", mae)
print("Root Mean Squared Error (RMSE):", rmse)
print("R2 Score:", r2)
```

```python
plt.figure(figsize=(10,6))
plt.plot(test_df["time"], y_test, label="Gerçek VG Temp", marker="o")
plt.plot(test_df["time"], y_pred, label="Tahmin Edilen VG Temp", marker="x")
plt.xlabel("Zaman")
plt.ylabel("VG Temp (BRS.T35_T_GBX_Avg)")
plt.title("Zamana Göre Gerçek ve Tahmin Edilen VG Temp Değerleri")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
